```python
from pathlib import Path
from PIL import Image
import numpy as np

image_path = Path("macbook.jpeg")
img = Image.open(image_path).convert("RGB")
# img = Image.open(image_path)

print("Image file:", image_path)
print("Image mode:", img.mode)
print("Image size (width, height):", img.size)

# Show the real image
img
```

#  Convert the real image to a NumPy array

```python

img_array = np.array(img)

print("Original image array shape:", img_array.shape)
print("Array dtype:", img_array.dtype)
```

```python
small_img = img.resize((4, 4))
small_array = np.array(small_img)

print("Small image shape:", small_array.shape)

# Show the small image
small_img
```

```python
print(small_array)
```

# Print all pixels one by one with their positions


```python
for row in range(small_array.shape[0]):
    for col in range(small_array.shape[1]):
        print(f"Pixel at ({row}, {col}) = {small_array[row, col]}")
```

# Decode the pixel matrix back into an image

```python
decoded_img = Image.fromarray(small_array.astype(np.uint8), mode="RGB")

# Show the decoded image
decoded_img
```

```python
print(decoded_img.size)
```
